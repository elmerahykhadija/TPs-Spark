```python
nombres = sc.parallelize([7, 10, 12, 17, 19], 2)
```

```python
carres=nombres.map(lambda n:n*n)
```

```python
carres.collect()
```

```python
print(nombres.count())
print(carres.count())
```

```python
couples=nombres.map(lambda n : (n//2,n%2))
```

```python
couples.collect()
```

```python
pairs=couples.filter(lambda n:n[1]==0)
```

```python
pairs.collect()
```

```python
mots=txt.flatMap(lambda l:l.lower().split(" "))
```

```python
print(mots.count())
```

```python
unique=txt.distinct()
```

```python
unique.count()
```

```python
ocurrence=mots.map(lambda w: (w, 1))
```

```python
ocurrence.collect()
```

```python
unique2=ocurrence.reduceByKey(lambda a, b: a + b)
```

```python
unique2.collect()
```

```python
unique2.sortBy(lambda kv: -kv[1]).take(5)
```

```python
unique2.takeOrdered(5, key=lambda kv: -kv[1])
```

```python
txt.filter(lambda l:"Spark" in l).count()
```

```python
wc = mots.count()

sc.parallelize([wc]).coalesce(1).saveAsTextFile(DATA + "/wc_out")
```
